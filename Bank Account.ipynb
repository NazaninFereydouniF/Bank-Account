```python
#Create a bank account to collect information about owners and their balance amount
class BankAccount:
    def __init__(self, owner, balance  = 0):
        self.owner = owner
        self.balance = balance

#Show information of each account belong to every owner
    def __str__(self):
        return f"Account Owner: {self.owner} \nAccount Balance: {self.balance}"

#Check every amount of deposition
    def deposit(self, dp_amt):
        self.balance += dp_amt
        print(f'Deposit Accepted. \nAmount of current deposit: {dp_amt} \nCurrent Account Balance: {self.balance}')

#check every withdrawal
#considering the account balance to allow withdraw
    def withdraw(self, wd_amt):
        if self.balance >= wd_amt:
            self.balance -= wd_amt
            print(f'Withdrawal Accepted. \nAmount of current withdraw: {wd_amt} \nCurrent Account Balance: {self.balance}')
        else:
            print('Funding is Unavailable!')
```

```python
act1 = BankAccount('John', 100)
act2 = BankAccount('Sara', 150)
act3 = BankAccount('Tom', 0)
act4 = BankAccount('Lisa', 200)
act5 = BankAccount('Amy', 50)
```

```python
act1.deposit(230)
act2.deposit(450)
act3.deposit(650)
act4.deposit(120)
act5.deposit(260)
```

```python
act1.withdraw(75)
act2.withdraw(65)
act3.withdraw(360)
act4.withdraw(500)
act5.withdraw(35)
```
